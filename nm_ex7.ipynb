```python
from diffusers import StableDiffusionPipeline
import torch
import os

# Initialize the model
pipe = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16)
pipe.to("cuda")

# Function to generate an image from a prompt
def generate_frame(prompt, frame_number, output_path):
    image = pipe(prompt).images[0]
    image.save(os.path.join(output_path, f"frame_{frame_number}.png"))
    print(f"Saved frame {frame_number}")

# Generate multiple frames (e.g., 30 frames for 1 second of video at 30 FPS)
output_folder = "generated_frames"
os.makedirs(output_folder, exist_ok=True)

prompt = "a couple dancing in the rain"
num_frames = 10

for i in range(num_frames):
    generate_frame(prompt, i, output_folder)
```

```python
import cv2
import os

def create_video_from_frames(image_folder, output_video_name, frame_rate=30, size=(800, 600)):
    images = []

    # Sort frames in the correct order
    for filename in sorted(os.listdir(image_folder)):
        if filename.endswith('.png') or filename.endswith('.jpg'):
            img_path = os.path.join(image_folder, filename)
            img = cv2.imread(img_path)
            img = cv2.resize(img, size)  # Resize images to the video frame size
            images.append(img)

    # Create the video writer
    fourcc = cv2.VideoWriter_fourcc(*'XVID')
    video_out = cv2.VideoWriter(output_video_name, fourcc, frame_rate, size)

    # Write frames to the video
    for image in images:
        video_out.write(image)

    video_out.release()
    print(f"Video saved as {output_video_name}")

# Example usage: create a video from generated frames
create_video_from_frames("generated_frames", "output_video.avi")
```
