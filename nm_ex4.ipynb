```python
from transformers import pipeline

# Load a zero-shot classification model
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

# Define categories relevant to resumes
resume_text = """
Experienced software engineer with 5 years in AI and Machine Learning.
Proficient in Python, TensorFlow, and OpenCV.
Worked on computer vision projects involving object detection and segmentation.
"""

categories = ["Software Development", "Data Science", "Machine Learning", "Project Management", "Marketing"]

# Analyze the resume
result = classifier(resume_text, candidate_labels=categories)

print("Predicted Job Category:", result["labels"][0])
print("Confidence Score:", result["scores"][0])
```

```python
import pdfplumber
from transformers import pipeline

# Load Hugging Face summarization model
summarizer = pipeline("summarization", model="t5-small")

# Function to extract text from PDF
def extract_text_from_pdf(pdf_path):
    with pdfplumber.open(pdf_path) as pdf:
        text = "\n".join([page.extract_text() for page in pdf.pages if page.extract_text()])
    return text.strip() if text else "No text extracted from the PDF."

# Function to analyze the resume
def analyze_resume(pdf_path):
    resume_text = extract_text_from_pdf(pdf_path)
    if "No text" in resume_text:
        return resume_text  # Return error if extraction failed

    # Summarize resume content
    summary = summarizer(resume_text, max_length=300, min_length=100, do_sample=False)

    feedback = f"""
    📌 **Resume Analysis:**
    - **Key Strengths:** {summary[0]['summary_text']}
    - **Areas for Improvement:** Improve ATS-friendliness by optimizing keywords and structure.
    - **Overall Score:** 7.5/10 (Based on clarity, structure, and readability)
    """
    return feedback

# Analyze the uploaded PDF
feedback = analyze_resume(pdf_path)
print(feedback)
```

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

# Load AI Model from Hugging Face
MODEL_NAME = "microsoft/DialoGPT-medium"  # Using 'medium' for better responses
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)

# Conversation Memory
chat_history_ids = None

# Function to generate chatbot response
def chatbot_response(user_input):
    global chat_history_ids

    # Encode user input
    new_input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors="pt")

    # Append new input to chat history
    bot_input_ids = torch.cat([chat_history_ids, new_input_ids], dim=-1) if chat_history_ids is not None else new_input_ids

    # Generate response with better randomness
    with torch.no_grad():
        chat_history_ids = model.generate(
            bot_input_ids,
            max_length=1000,
            pad_token_id=tokenizer.eos_token_id,
            do_sample=True,  # Enables randomness
            top_k=50,  # Chooses from top 50 tokens
            temperature=0.7  # Adjusts randomness level
        )

    # Decode response
    bot_reply = tokenizer.decode(chat_history_ids[:, bot_input_ids.shape[-1]:][0], skip_special_tokens=True)
    return bot_reply

# Chat Loop
while True:
    user_msg = input("You: ")
    if user_msg.lower() in ["exit", "bye"]:
        print("Bot: Bye! Have a great day! 😊")
        break
    bot_response = chatbot_response(user_msg)
    print(f"Bot: {bot_response}")
```

```python
from transformers import pipeline

# Load Hugging Face summarization model
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

def chatbot():
    print("📝 Emily: Your summarization assistant! (Type 'exit' to quit)\n")
    while True:
        user_input = input("You: ")
        if user_input.lower() == "exit":
            print("Emily: Bye! Have a great day! 😊")
            break
        summary = summarizer(user_input, max_length=50, min_length=10, do_sample=False)
        print("Emily:", summary[0]['summary_text'])

# Start chatbot
chatbot()
```

```python
import langid
from googletrans import Translator

def detect_language(text):
    """Detects the language of the input text."""
    return langid.classify(text)[0]  # Returns language code (e.g., 'en', 'ta', 'hi')

def translate_text(text, target_lang):
    """Translates the text into the target language."""
    translator = Translator()
    translation = translator.translate(text, dest=target_lang)
    return translation.text

def main():
    print("🔹 AI-Powered Language Translator 🔹")
    print("Type 'exit' to quit.\n")

    while True:
        text = input("Enter text to translate (or type 'exit' to quit): ")
        if text.lower() == "exit":
            print("Goodbye! 👋")
            break

        detected_lang = detect_language(text)
        print(f"\nDetected Language: {detected_lang}")

        target_lang = input("Enter target language code (e.g., 'en' for English, 'fr' for French, 'es' for Spanish): ")

        translated_text = translate_text(text, target_lang)
        print(f"Translated Text ({target_lang}): {translated_text}\n")

if __name__ == "__main__":
    main()
```
