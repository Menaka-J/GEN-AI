```python
import gradio as gr
from transformers import pipeline

# Load a pre-trained text classification model
classifier = pipeline("text-classification")

# Define the function to classify text
def classify_text(text):
    results = classifier(text)
    return results[0]['label'], results[0]['score']  # Return label and score

# Create a Gradio interface
interface = gr.Interface(
    fn=classify_text,  # Function to handle user input
    inputs="text",      # Input type (text box)
    outputs=["label", "number"],  # Output types (label and score)
    title="Text Classification",
    description="Enter some text and classify it into categories."
)

# Launch the interface
interface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Load a pre-trained sentiment analysis model
sentiment_analyzer = pipeline("sentiment-analysis")

# Define the function to analyze sentiment
def analyze_sentiment(text):
    results = sentiment_analyzer(text)
    return results[0]['label'], results[0]['score']  # Return label and score

# Create a Gradio interface
interface = gr.Interface(
    fn=analyze_sentiment,  # Function to handle user input
    inputs="text",         # Input type (text box)
    outputs=["label", "number"],  # Output types (label and score)
    title="Sentiment Analysis",
    description="Enter some text to analyze its sentiment."
)

# Launch the interface
interface.launch()
```

```python
import gradio as gr
import spacy

# Load the English NLP model from spaCy
nlp = spacy.load("en_core_web_sm")

# Define the function for POS tagging
def pos_tagging(text):
    doc = nlp(text)
    # Create a list of tuples containing (word, POS tag)
    pos_tags = [(token.text, token.pos_) for token in doc]
    return pos_tags

# Create a Gradio interface
interface = gr.Interface(
    fn=pos_tagging,  # Function to handle user input
    inputs="text",    # Input type (text box)
    outputs="json",   # Output type (JSON format for structured output)
    title="Part-of-Speech Tagging",
    description="Enter a sentence to see the part-of-speech tags for each word."
)

# Launch the interface
interface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Load the translation model (e.g., English to French)
translator = pipeline("translation_en_to_fr")  # Change model based on your needs

# Define the function for translation
def translate_text(text):
    translation = translator(text, max_length=40)  # Adjust max_length as needed
    return translation[0]['translation_text']

# Create a Gradio interface
interface = gr.Interface(
    fn=translate_text,  # Function to handle user input
    inputs="text",      # Input type (text box)
    outputs="text",     # Output type (text box)
    title="Machine Translation",
    description="Enter text in English to translate it into French."
)

# Launch the interface
interface.launch()
```
