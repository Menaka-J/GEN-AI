```python
from transformers import MarianMTModel, MarianTokenizer

def translate_text(text, src_lang='de', tgt_lang='en'):
    # Define the model and tokenizer
    model_name = f'Helsinki-NLP/opus-mt-{src_lang}-{tgt_lang}'
    model = MarianMTModel.from_pretrained(model_name)
    tokenizer = MarianTokenizer.from_pretrained(model_name)

    # Tokenize the input text
    tokens = tokenizer(text, return_tensors="pt", padding=True, truncation=True)

    # Perform translation
    translated_tokens = model.generate(**tokens)

    # Decode the translated tokens to get the translated text
    translated_text = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)

    return translated_text

if __name__ == "__main__":
    # Get user input for the text to be translated
    text = input("Enter the text to translate: ")

    # Get the source and target language from the user
    src_lang = input("Enter the source language code (default 'de' for German): ") or 'de'
    tgt_lang = input("Enter the target language code (default 'en' for English): ") or 'en'

    # Translate the text
    translated_text = translate_text(text, src_lang, tgt_lang)

    print(f"Translated text: {translated_text}")

```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

def generate_text(prompt, model_name="gpt2", max_length=100):
    # Load pre-trained model and tokenizer from Hugging Face
    tokenizer = GPT2Tokenizer.from_pretrained(model_name)
    model = GPT2LMHeadModel.from_pretrained(model_name)

    # Encode the input prompt into tokens
    input_ids = tokenizer.encode(prompt, return_tensors="pt")

    # Generate text based on the input prompt
    output = model.generate(input_ids, max_length=max_length, num_return_sequences=1, no_repeat_ngram_size=2, pad_token_id=tokenizer.eos_token_id)

    # Decode the generated tokens back to a string
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)

    return generated_text

if __name__ == "__main__":
    # Get user input for the prompt
    prompt = input("Enter a prompt to generate text: ")

    # Generate text using GPT-2
    generated_text = generate_text(prompt)

    # Output the generated text
    print(f"Generated Text: {generated_text}")

```

```python
# Example dictionary for translation from English to Tamil
# You can expand this dictionary with more words for better translation
translation_dict = {
    'hello': 'வணக்கம்',
    'how': 'எப்படி',
    'are': 'இருக்கின்றீர்கள்',
    'you': 'நீங்கள்',
    'good': 'நல்ல',
    'morning': 'பொழுது',
    'i': 'நான்',
    'am': 'ஆகின்றேன்',
    'fine': 'நல்ல',
    'thank': 'நன்றி',
    'please': 'தயவுசெய்து'
}

def translate_text(text, src_lang='en', tgt_lang='ta'):
    # Split the text into individual words
    words = text.lower().split()

    # Translate each word using the dictionary
    translated_words = []
    for word in words:
        translated_word = translation_dict.get(word, word)  # Default to original word if not in dict
        translated_words.append(translated_word)

    # Join the translated words back into a sentence
    translated_text = ' '.join(translated_words)
    return translated_text

if __name__ == "__main__":
    # Get user input for the text to be translated
    text = input("Enter the text to translate: ")

    # Translate the text (English to Tamil by default)
    translated_text = translate_text(text)

    print(f"Translated text: {translated_text}")
```

```python
from gtts import gTTS
from IPython.display import Audio, display

def text_to_speech(text, language='en'):

    tts = gTTS(text=text, lang=language, slow=False)

    tts.save("output.mp3")

    import os
    if os.path.exists("output.mp3"):
        print("Audio file saved successfully.")
    else:
        print("Error: Audio file not saved.")

    return Audio("output.mp3", autoplay=True)

if __name__ == "__main__":
    text = input("Enter text to convert to speech: ")
    audio = text_to_speech(text)

    display(audio)
```
